# 01 Data preparation

This notebook documents the 510300 Garman-Klass variance input used by all later experiments.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

REPO_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from src.volatility import load_rv_series, validate_split

FIG_DIR = REPO_ROOT / "results" / "figures"
TABLE_DIR = REPO_ROOT / "results" / "tables"
FIG_DIR.mkdir(parents=True, exist_ok=True)
TABLE_DIR.mkdir(parents=True, exist_ok=True)
plt.rcParams["axes.unicode_minus"] = False
```

## Load the frozen research series

The repository keeps the exact 404-day series used in the reported results. Network download is optional and is not used when the CSV is present.

```python
rv, source = load_rv_series(
    REPO_ROOT / "data" / "rv_series.csv",
    allow_network=False,
    allow_synthetic_fallback=False,
)
train_end_idx, n_test = validate_split(rv)

print("source:", source)
print("observations:", len(rv))
print("range:", rv.index.min().date(), "to", rv.index.max().date())
print("train/test:", train_end_idx, n_test)
print("mean:", rv.mean())
print("std:", rv.std())
rv.head()
```

## Plot the variance series

```python
fig, ax = plt.subplots(figsize=(13, 4.5))
ax.plot(rv.index, rv.values, color="tab:blue", linewidth=1.1)
ax.axvline(pd.Timestamp("2024-02-29"), color="black", linestyle="--", linewidth=1)
ax.set_title("CSI 300 ETF 510300 Garman-Klass variance")
ax.set_xlabel("Date")
ax.set_ylabel("GK variance")
ax.grid(alpha=0.25)
fig.tight_layout()
fig.savefig(FIG_DIR / "rv_series.png", dpi=200, bbox_inches="tight")
plt.show()
```
